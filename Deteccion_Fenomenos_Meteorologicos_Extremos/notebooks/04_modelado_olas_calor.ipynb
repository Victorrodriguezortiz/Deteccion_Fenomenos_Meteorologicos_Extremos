```python
from src.modeling.dataset_preparation import generar_dataset_final_ml_limpio_v3, separar_X_y
from src.modeling.pipeline_modelos import ExtremeEventModelPipeline
```

```python
df_train, df_test, features, scaler = generar_dataset_final_ml_limpio_v3(
    ruta_nc="train/X_olas_calor_final.nc",
    var_target="mx2t",
    p_umbral=0.95,
    max_lags=30,
    dias_test=365,
)

X_train, y_train, X_test, y_test = separar_X_y(df_train, df_test, features)
```

```python
pipeline_olas = ExtremeEventModelPipeline(
    X_train=X_train,
    y_train=y_train,
    X_test=X_test,
    y_test=y_test,
    feature_names=features,
    threshold_metric="f2",
    beta=2.0,
)

pipeline_olas.train(
    models=[
        "Logistic Regression",
        "Random Forest",
        "Extra Trees",
        "HistGradientBoosting",
        "Neural Network",
    ],
    optimize=False,
)

summary_olas = pipeline_olas.get_summary_df()
summary_olas
```
